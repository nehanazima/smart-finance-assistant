````python
import pandas as pd

# 🤖 AI Collaboration: Data Loading and Cleaning Function
# Ask AI to help you create a robust data loading function

def load_and_clean_transaction_data(df):
    """
    Load and clean transaction data for the Smart Finance Assistant.
    Modified to accept a DataFrame directly for testing with `df_sample`.

    🤖 AI Collaboration Prompt:
    "Create a function to load CSV transaction data with Date, Amount, Category,
    Description columns. Handle dollar signs in Amount, missing values, and
    data validation. Include clear business-focused error messages."

    Args:
        df (pandas.DataFrame): The input DataFrame (or file_path if reading from CSV).
    Returns:
        pandas.DataFrame: Cleaned transaction data
    """
    # If the function is meant to read from a file path, uncomment and modify the following lines:
    # if isinstance(file_path, str):
    #     try:
    #         df = pd.read_csv(file_path)
    #     except FileNotFoundError:
    #         print(f"Error: File not found at {file_path}")
    #         return pd.DataFrame() # Return an empty DataFrame on error
    #     except Exception as e:
    #         print(f"Error reading CSV: {e}")
    #         return pd.DataFrame()
    # elif not isinstance(df, pd.DataFrame):
    #     print("Error: Input must be a file path or a pandas DataFrame.")
    #     return pd.DataFrame()

    if df.empty:
        print("Warning: Input DataFrame is empty. No data to process.")
        return df

    required_columns = ['Date', 'Amount', 'Category', 'Description']
    if not all(col in df.columns for col in required_columns):
        missing_cols = [col for col in required_columns if col not in df.columns]
        print(f"Error: Missing required columns: {', '.join(missing_cols)}")
        return pd.DataFrame() # Return empty DataFrame if critical columns are missing

    # Make a copy to avoid SettingWithCopyWarning
    cleaned_df = df.copy()

    # Step 1: Load and validate data & Handle missing or invalid data
    # Convert 'Date' column to datetime objects
    # Invalid dates will be set to NaT (Not a Time) and then dropped or flagged
    initial_rows = len(cleaned_df)
    cleaned_df['Date'] = pd.to_datetime(cleaned_df['Date'], errors='coerce')
    if cleaned_df['Date'].isnull().any():
        invalid_dates_count = cleaned_df['Date'].isnull().sum()
        print(f"Warning: {invalid_dates_count} transactions with invalid or missing dates found and will be dropped.")
        cleaned_df.dropna(subset=['Date'], inplace=True)

    # Handle 'Category' missing values
    if cleaned_df['Category'].isnull().any():
        print("Info: Missing categories found. Setting them to 'Uncategorized'.")
        cleaned_df['Category'].fillna('Uncategorized', inplace=True)

    # Remove duplicate transactions
    duplicates_count = cleaned_df.duplicated().sum()
    if duplicates_count > 0:
        print(f"Info: {duplicates_count} duplicate transactions found and removed.")
        cleaned_df.drop_duplicates(inplace=True)

    # Step 2: Clean financial data
    # Remove currency symbols ($) and commas from Amount column
    if 'Amount' in cleaned_df.columns:
        cleaned_df['Amount'] = cleaned_df['Amount'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
        # Convert Amount to numeric
        # Invalid amounts will be set to NaN (Not a Number)
        cleaned_df['Amount'] = pd.to_numeric(cleaned_df['Amount'], errors='coerce')

        # Handle non-numeric or missing amounts
        if cleaned_df['Amount'].isnull().any():
            invalid_amounts_count = cleaned_df['Amount'].isnull().sum()
            print(f"Warning: {invalid_amounts_count} transactions with invalid or missing amounts found and will be dropped.")
            cleaned_df.dropna(subset=['Amount'], inplace=True)

    if len(cleaned_df) < initial_rows:
        print(f"Info: {initial_rows - len(cleaned_df)} rows were dropped due to invalid or missing critical data.")

    # Separate positive spending from refunds /negative transactions (optional for cleaning, but good to have)
    # This can be used later for analysis, but for cleaning, we ensure the column is numeric.
    cleaned_df['Is_Refund'] = cleaned_df['Amount'] < 0

    print("✅ Data loaded and cleaned successfully.")
    return cleaned_df

# --- Sample Data Definition (Moved here for testing) ---
sample_transactions = {
    'Date': ['2024-08-01', '2024-08-02', '2024-08-03', '2024-08-04', '2024-08-05'],
    'Amount': ['$45.50', '$12.00', '$89.95', '$3.50', '-$25.00'],  # Note: includes $ signs and refund
    'Category': ['Groceries', 'Transport', 'Entertainment', 'Coffee', 'Refund'],
    'Description': ['Woolworths', 'Opal Card', 'Concert Tickets', 'Campus Cafe', 'Returned Item']
}
df_sample = pd.DataFrame(sample_transactions)

# Test your function
print("\n--- Testing with sample data ---")
clean_data = load_and_clean_transaction_data(df_sample)
print("\nCleaned Data Head:")
display(clean_data.head())
print("\nCleaned Data Info:")
clean_data.info()
````

````python
# 🤖 AI Collaboration: Spending Analysis Function
# Ask AI to help you create comprehensive spending analysis

def analyze_spending_patterns(df):
    """
    Analyze spending patterns and generate business insights

    🤖 AI Collaboration Prompt:
    "Create a function that analyzes spending by category, calculates percentages,
    identifies top spending areas, and generates actionable financial insights
    formatted for business presentation."

    Args:
        df: Cleaned transaction DataFrame
    Returns:
        dict: Analysis results and insights
    """
    if df.empty:
        return {"summary": "No transactions to analyze.", "categories": {}}

    analysis_results = {}

    # Calculate total spending (positive amounts only)
    total_spending = df[df['Amount'] > 0]['Amount'].sum()
    net_spending = df['Amount'].sum()
    total_transactions = len(df)

    analysis_results['Total Spending (Excluding Refunds)'] = f"${total_spending:,.2f}"
    analysis_results['Net Spending (Including Refunds)'] = f"${net_spending:,.2f}"
    analysis_results['Total Transactions'] = total_transactions

    # Group by category for detailed analysis
    category_summary = df.groupby('Category')['Amount'].agg(['sum', 'count', 'mean']).reset_index()
    category_summary.columns = ['Category', 'Total Amount', 'Transaction Count', 'Average Amount']

    # Calculate percentage of total spending for each category (positive spending only)
    # Filter for positive spending when calculating percentages
    positive_spending_df = df[df['Amount'] > 0]
    category_positive_summary = positive_spending_df.groupby('Category')['Amount'].sum().reset_index()
    category_positive_summary.columns = ['Category', 'Positive Total Amount']

    if total_spending > 0:
        category_positive_summary['Percentage'] = (category_positive_summary['Positive Total Amount'] / total_spending * 100).map('{:.2f}%'.format)
    else:
        category_positive_summary['Percentage'] = '0.00%'

    # Merge percentage back to main category summary
    category_summary = pd.merge(category_summary, category_positive_summary[['Category', 'Percentage']], on='Category', how='left')
    category_summary['Percentage'].fillna('0.00%', inplace=True)

    # Format for business presentation
    category_summary['Total Amount'] = category_summary['Total Amount'].map('${:,.2f}'.format)
    category_summary['Average Amount'] = category_summary['Average Amount'].map('${:,.2f}'.format)

    analysis_results['Category Spending Summary'] = category_summary.to_dict(orient='records')

    # Identify top spending categories (excluding refunds/negative amounts for this insight)
    top_spending_categories = category_positive_summary.sort_values(by='Positive Total Amount', ascending=False).head(3)
    analysis_results['Top 3 Spending Categories'] = top_spending_categories[['Category', 'Positive Total Amount', 'Percentage']].to_dict(orient='records')

    # Detect unusually large transactions (simple heuristic: e.g., > 3 * overall average positive transaction)
    if positive_spending_df['Amount'].mean() > 0:
        average_positive_transaction = positive_spending_df['Amount'].mean()
        large_transactions = df[df['Amount'] > (3 * average_positive_transaction)]
        if not large_transactions.empty:
            analysis_results['Potentially Large Transactions'] = large_transactions[['Date', 'Amount', 'Category', 'Description']].to_dict(orient='records')
        else:
            analysis_results['Potentially Large Transactions'] = "No unusually large transactions identified based on heuristic."
    else:
        analysis_results['Potentially Large Transactions'] = "Cannot determine large transactions: no positive spending."

    print("✅ Spending patterns analyzed successfully.")
    return analysis_results

# Test your function
print("\n--- Testing spending analysis with cleaned data ---")
analysis = analyze_spending_patterns(clean_data)

# Print the analysis results in a structured way for readability
print("\nFinancial Analysis Report:")
for key, value in analysis.items():
    print(f"\n{key}:")
    if isinstance(value, list):
        for item in value:
            print(f"  {item}")
    elif isinstance(value, dict):
        for sub_key, sub_value in value.items():
            print(f"  {sub_key}: {sub_value}")
    else:
        print(f"  {value}")
````

````python

def generate_financial_recommendations(analysis_data):
    """
    Generate actionable financial recommendations based on spending analysis
    """
    recommendations = []

    # Step 1: Summarize spending by category
    # The input `analysis_data` is expected to be the 'clean_data' DataFrame,
    # not the dictionary output from `analyze_spending_patterns`.
    # Ensure 'Amount' column is numeric for sum operations.

    # We need to use the actual cleaned DataFrame here, not the analysis dictionary.
    # Assuming analysis_data is the `clean_data` DataFrame.

    # Filter for positive spending for summary to avoid refunds skewing percentages
    positive_spending_df = analysis_data[analysis_data['Amount'] > 0]
    total_positive_spending = positive_spending_df['Amount'].sum()

    if total_positive_spending == 0:
        recommendations.append("No positive spending recorded for analysis.")
        return "\n".join(recommendations)

    category_summary = positive_spending_df.groupby('Category')['Amount'].sum().reset_index()
    category_summary['Percentage'] = (category_summary['Amount'] / total_positive_spending * 100).round(2)

    # Step 2: Identify top spending categories
    top_categories = category_summary.sort_values('Amount', ascending=False).head(3)

    for _, row in top_categories.iterrows():
        recommendations.append(
            f"Category '{row['Category']}' accounts for {row['Percentage']:.2f}% of your spending." # Fixed f-string
            f" Consider reviewing your spending in this category and setting a budget."
        )

    # Step 3: Detect large one-off transactions
    # Using a fixed threshold for 'large' for clearer demonstration and testability.
    large_transaction_threshold = 200.0 # Example fixed threshold for a 'large' transaction
    large_transactions = analysis_data[analysis_data['Amount'] > large_transaction_threshold]

    if not large_transactions.empty:
        for _, row in large_transactions.iterrows():
            recommendations.append(
                f"Large transaction of ${row['Amount']:.2f} detected in '{row['Category']}' " # Fixed column access
                f"on {row['Date'].strftime('%Y-%m-%d')}. Ensure this aligns with your budget."
            )

    # Step 4: Highlight refunds
    refunds = analysis_data[analysis_data['Is_Refund'] == True]
    if not refunds.empty:
        for _, row in refunds.iterrows():
            recommendations.append(
                f"Refund of ${abs(row['Amount']):.2f} detected in '{row['Category']}' "
                f"on {row['Date'].strftime('%Y-%m-%d')}."
            )
    else:
        recommendations.append("No refunds detected in this period.")

    # Step 5: Return formatted report
    return "\n".join(recommendations)

# Test your function with the 'clean_data' DataFrame
print("\n--- Generating Financial Recommendations ---")
recommendations_report = generate_financial_recommendations(clean_data)
print(recommendations_report)
````

````python
import os
# from hands_on_ai.chat import get_response # Removed for offline functionality

def create_finance_chat_personality():
    """
    Set up a financial advisor chat personality. This version works offline.

    🤖 AI Collaboration Prompt:
    "Help me create a system prompt for a friendly, professional financial
    advisor chatbot that can provide spending advice based on transaction
    analysis. The personality should be encouraging but practical."
    """
    # Define personality parameters
    name = "Smart Finance Advisor"
    traits = ["friendly", "encouraging", "practical"]
    knowledge_focus = ["spending habits", "budgeting", "category trends", "savings tips"]
    response_style = "actionable, concise, user-friendly"

    # This nested function captures the personality in its closure
    def finance_chat_say(prompt):
        persona_str = f"You are a helpful and {', '.join(traits)} financial advisor named {name}."
        persona_str += f" Your expertise focuses on {', '.join(knowledge_focus)}. "
        persona_str += f" Your response style should be {response_style}."

        # Simple rule-based response for offline functionality
        lower_prompt = prompt.lower()
        if "spending" in lower_prompt or "budget" in lower_prompt:
            response = "That's a great question about managing your finances! Let's look at your spending habits and find practical ways to optimize your budget. Remember, even small changes can make a big difference!"
        elif "save" in lower_prompt or "savings" in lower_prompt:
            response = "Saving is a fantastic goal! We can explore some actionable strategies to boost your savings, whether it's through identifying areas to cut back or setting clear financial milestones."
        elif "recommendation" in lower_prompt or "advice" in lower_prompt:
            response = "I'm here to provide actionable recommendations! Please tell me more about what you'd like advice on, and I'll offer some practical, user-friendly insights."
        else:
            response = "Hello there! I'm your Smart Finance Advisor. How can I help you understand and improve your financial situation today? Feel free to ask me anything about your spending, budget, or savings!"

        return f"{name}: {response}"

    return finance_chat_say # Return the callable function

# Example usage:
finance_chat = create_finance_chat_personality() # finance_chat is now a function
response = finance_chat("I spent too much on coffee this week. What should I do?") # Call it directly
print(response)

response_2 = finance_chat("I want to save more money. Any tips?")
print(response_2)

response_3 = finance_chat("Tell me a joke.")
print(response_3)
````
